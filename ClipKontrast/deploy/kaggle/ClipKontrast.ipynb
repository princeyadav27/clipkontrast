```python
from pathlib import Path
import sys
from kaggle_secrets import UserSecretsClient

PROJECT = Path('/kaggle/input/clipkontrast/ClipKontrast')  # folder containing requirements.txt
if not (PROJECT / 'deploy/kaggle/launch.py').is_file():
    raise FileNotFoundError('Set PROJECT to the attached ClipKontrast code folder containing deploy/kaggle/launch.py.')
sys.path.insert(0, str(PROJECT))
from deploy.kaggle.launch import start
secrets = UserSecretsClient()
backup = {}  # default: no external storage
USE_S3 = False  # set True after adding/granting the secrets below
if USE_S3:
    for name in ('CLIPKONTRAST_BACKUP_BUCKET', 'AWS_DEFAULT_REGION', 'AWS_ACCESS_KEY_ID', 'AWS_SECRET_ACCESS_KEY'):
        backup[name] = secrets.get_secret(name)
    backup['CLIPKONTRAST_BACKUP_PREFIX'] = 'clipkontrast-private'
    backup['CLIPKONTRAST_BACKUP_AUTO'] = '1'
    # For temporary AWS credentials, also load AWS_SESSION_TOKEN from Secrets.
start(secrets.get_secret('CLIPKONTRAST_NOTEBOOK_PASSWORD'), backup_env=backup)
```
